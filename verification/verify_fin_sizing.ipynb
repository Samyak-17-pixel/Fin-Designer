# verify_fin_sizing

Geometry equations EQ-GEO-* vs production sizing.

```python
import math
from auv_fin_design.domain.geometry.sizing import size_fin, mean_aerodynamic_chord

g = size_fin(10.0, 1123.0, cl=0.2, aspect_ratio=0.8, taper_ratio=0.5, thickness_ratio=0.18)
S = 10.0 / (1123.0 * 0.2)
b = math.sqrt(S * 0.8)
assert abs(g.area - S) / S < 0.01
assert abs(g.span - b) / b < 0.01
mac = mean_aerodynamic_chord(g.root_chord, 0.5)
assert abs(g.mac - mac) / mac < 0.01
print('area', g.area, 'span', g.span, 'mac', g.mac)
print('PASS')
```
