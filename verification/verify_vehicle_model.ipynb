## Engineering Theory

Uniform circular cylinder hull (V1):

$$
R = D/2,\quad A_f = \pi R^2,\quad S_w = \pi D L,\quad \nabla = \pi R^2 L
$$

Solid-cylinder inertia about CG:

$$
I_x = \tfrac12 m R^2,\quad I_y = I_z = \tfrac{1}{12} m (3R^2 + L^2)
$$

Aft fins: root LE at $0.92 L$ from nose; CG at mid-length.

```python
import math
from auv_fin_design.domain.vehicle.model import VehicleModel

L, D, m = 1.35, 0.1685, 24.0
rho, nu = 998.2, 1.004e-6

# Manual
R = D / 2
Af = math.pi * R**2
Sw = math.pi * D * L
Vol = math.pi * R**2 * L
Ix = 0.5 * m * R**2
Iz = (1/12) * m * (3*R**2 + L**2)
x_cg = 0.5 * L
x_fin = 0.92 * L

print('Manual R, Af, Sw, Vol, Ix, Iz, x_cg, x_fin:')
print(R, Af, Sw, Vol, Ix, Iz, x_cg, x_fin)

# Production
v = VehicleModel(length=L, diameter=D, mass=m, water='freshwater')
checks = {
    'R': (R, v.radius),
    'Af': (Af, v.frontal_area),
    'Sw': (Sw, v.wetted_area),
    'Vol': (Vol, v.volume),
    'Ix': (Ix, v.Ix),
    'Iz': (Iz, v.Iz),
    'x_cg': (x_cg, v.x_cg),
    'x_fin': (x_fin, v.x_fin_root_le),
}
tol = 0.01
for name, (manual, prod) in checks.items():
    err = abs(manual - prod) / abs(manual)
    status = 'PASS' if err <= tol else 'FAIL'
    print(f'{name}: rel_err={err:.3e} [{status}]')
```

## Conclusions

Vehicle model matches analytical cylinder formulas within 1% tolerance.  
Limitation: solid inertia and cylinder wetted area (no nose/tail) are intentional V1 simplifications.
